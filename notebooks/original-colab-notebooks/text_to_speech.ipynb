```python

%%writefile app.py

# PART 1: Imports and Page Setup

import streamlit as st
from gtts import gTTS
from io import BytesIO

# Set browser tab title and icon
st.set_page_config(
    page_title="Text to Speech",
    page_icon="🔊"
)

st.title("🔊 Text to Speech")

# Multi-line text input
text = st.text_area("Enter your text:", height=150)


# PART 2: Language and Speed Options

languages = {
    "English": "en",
    "Hindi": "hi",
    "Kannada": "kn",
    "Marathi": "mr",
    "Tamil": "ta",
    "Telugu": "te"
}

# Select language
lang_name = st.selectbox(
    "Language:",
    list(languages.keys())
)

# Select speech speed
slow = st.checkbox("Speak slowly")

# Debug line (disabled)
# print(f"Selected language: {lang_name}, Slow: {slow}")


# PART 3: Button and Audio Generation

if st.button("Convert to Speech"):

    # Check whether text is empty
    if not text.strip():
        st.warning("Please enter some text first!")

    else:
        # Show loading spinner
        with st.spinner("Generating audio..."):

            try:
                # Build the speech
                tts = gTTS(
                    text=text,
                    lang=languages[lang_name],
                    slow=slow
                )

                # Create an empty in-memory file
                audio = BytesIO()

                # Write MP3 bytes into memory
                tts.write_to_fp(audio)

                # Rewind to the start
                audio.seek(0)


                # PART 4: Play and Download Audio

                # Show audio player
                st.audio(audio, format="audio/mp3")

                # Show download button
                st.download_button(
                    label="Download MP3",
                    data=audio.getvalue(),
                    file_name="speech.mp3",
                    mime="audio/mpeg"
                )

            except Exception as e:
                # Show error message
                st.error(f"Something went wrong: {e}")
```
