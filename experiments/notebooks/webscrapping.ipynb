```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url="https://webscraper.io/test-sites/e-commerce/allinone/phones"

response = requests.get(url)
```

```python
response.status_code
```

```python
print(response.text)
```

```python
if response.status_code==200:
  soup=BeautifulSoup(response.text,"html.parser")
  products=soup.find_all('div',class_='product-wrapper card-body')
  product_data=[]
for product in products:
  try:
    name=product.find('a',class_='title').get_text(strip=True)
    price=product.find('h4',class_='price').get_text(strip=True)
    review=product.find('p',class_='description').get_text(strip=True)
    product_data.append({'Name':name,'Price':price,'Review':review})
  except Exception as e:
    print(f"Error extracting product data: {e}")
```

```python

if response.status_code == 200:
    df = pd.DataFrame(product_data)
    df.to_csv('tablets_data.csv', index=False)
    print("Tablet details scraped and saved to tablets_data.csv")
else:
    print("Failed to retrieve the webpage. Status code:", response.status_code)
```
